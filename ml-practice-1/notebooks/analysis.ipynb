```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем датасет (так как блокнот в папке notebooks, выходим на уровень выше в data)
df = pd.read_csv('../data/Iris.csv')

# 1. Размер таблицы (строки, столбцы)
print("Размер датасета (строк, столбцов):", df.shape)

# 2. Типы столбцов
print("\nТипы данных:\n", df.dtypes)

# 3. Количество пропусков по столбцам
print("\nПропуски по столбцам:\n", df.isnull().sum())

# 4. Базовые статистики числовых признаков
print("\nБазовые статистики:")
display(df.describe())
```

```python
# График 1: Распределение длины чашелистика (числовой признак)
plt.figure(figsize=(8, 5))
sns.histplot(df['SepalLengthCm'].dropna(), bins=20, kde=True, color='teal')
plt.title('Распределение длины чашелистика (SepalLengthCm)')
plt.xlabel('Длина чашелистика (см)')
plt.ylabel('Частота')
plt.show()

# График 2: Соотношение категорий (количество образцов каждого вида)
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Species', palette='Set2')
plt.title('Количество образцов каждого вида ирисов')
plt.xlabel('Вид')
plt.ylabel('Количество')
plt.show()
```

### Выводы по датасету Iris:
1. **Отсутствие пропусков:** Анализ показал, что датасет полностью заполнен, пропущенных значений (`null`) в строках и столбцах нет, что упрощает дальнейшую работу.
2. **Структура признаков:** Все основные признаки (длина и ширина чашелистика, длина и ширина лепестка) являются числовыми (с плавающей точкой), в то время как целевая переменная `Species` представлена в виде категорий (видов растений).
3. **Сбалансированность классов:** На графике распределения видов видно, что выборка сбалансирована — каждый из трех видов представлен равным количеством образцов.
